This notebook combines the petal and sepal measurements and calculates the requested statistics for the full iris dataset.

## Import Pandas and Load the Data

```python
import pandas as pd

petal_data = pd.read_csv('Petal_Data.csv')
sepal_data = pd.read_csv('Sepal_Data.csv')

print('Petal data shape:', petal_data.shape)
print('Sepal data shape:', sepal_data.shape)
```

## 1. Combine the Datasets
The extra index columns are removed, and the files are merged using both `sample_id` and `species`. This keeps the measurements attached to the correct flower even though the source files use different row orders.

```python
petal_data = petal_data.drop(columns=['Unnamed: 0'])
sepal_data = sepal_data.drop(columns=['Unnamed: 0'])

combined_data = pd.merge(
    petal_data,
    sepal_data,
    on=['sample_id', 'species'],
    how='inner',
    validate='one_to_one'
)

combined_data = combined_data[[
    'sample_id', 'species', 'petal_length', 'petal_width',
    'sepal_length', 'sepal_width'
]]

print('Combined data shape:', combined_data.shape)
combined_data.head(10)
```

```python
print('Missing values by column:')
print(combined_data.isna().sum())

print('Samples by species:')
print(combined_data['species'].value_counts())
```

## 2. Correlations Between Each Measurement
Four measurement variables create six unique pairwise comparisons.

```python
correlations = {
    'petal_length vs petal_width': combined_data['petal_length'].corr(combined_data['petal_width']),
    'petal_length vs sepal_length': combined_data['petal_length'].corr(combined_data['sepal_length']),
    'petal_length vs sepal_width': combined_data['petal_length'].corr(combined_data['sepal_width']),
    'petal_width vs sepal_length': combined_data['petal_width'].corr(combined_data['sepal_length']),
    'petal_width vs sepal_width': combined_data['petal_width'].corr(combined_data['sepal_width']),
    'sepal_length vs sepal_width': combined_data['sepal_length'].corr(combined_data['sepal_width'])
}

correlation_results = pd.Series(correlations, name='Correlation')
correlation_results.round(4)
```

## 3. Average of Each Measurement

```python
measurement_columns = [
    'petal_length', 'petal_width', 'sepal_length', 'sepal_width'
]

average_results = combined_data[measurement_columns].mean()
average_results.round(4)
```

## 4. Median of Each Measurement

```python
median_results = combined_data[measurement_columns].median()
median_results.round(4)
```

## 5. Standard Deviation of Each Measurement

```python
standard_deviation_results = combined_data[measurement_columns].std()
standard_deviation_results.round(4)
```

## Summary of Descriptive Statistics

```python
summary_results = pd.DataFrame({
    'Average': average_results,
    'Median': median_results,
    'Standard Deviation': standard_deviation_results
})

summary_results.round(4)
```
